```python
# Меняем названия полей
df.rename(columns={'SeriousDlqin2yrs':'Прогноз просрочки в течении 2 лет',
                   'RevolvingUtilizationOfUnsecuredLines':'Остаток на кред.картах, %',
                   'age':'Возраст',
                   'NumberOfTime30-59DaysPastDueNotWorse':'Кол-во просрочек 30-59 дней',
                   'DebtRatio':'Платежей от долга, % или абс.платеж',
                   'MonthlyIncome':'Месячный доход',
                   'NumberOfOpenCreditLinesAndLoans':'Кол-во кредитов',
                   'NumberOfTimes90DaysLate':'Кол-во просрочек на 90 дней и более',
                   'NumberRealEstateLoansOrLines':'Кол-во ипотек',
                   'NumberOfTime60-89DaysPastDueNotWorse':'Кол-во сред.просрочек (60-89 дней) за 2 года',
                   'NumberOfDependents':'Кол-во иждевенцев в семье'}, inplace=True)
df
```

```python
df[df['Месячный доход'].isna() == False].head(2)
# ПОХОЖЕ В ОПИСАНИИ ПОЛЕЙ 'Остаток на кред.картах, %' и 'Платежей от долга, % или абс.платеж' ОШИБКА
# величины скорее всего выраженны в долях, а не % !!!
```

```python
# Корректируем "Платежей от долга..."
df['Платежей от долга, % или абс.платеж'] = df[df['Месячный доход'].isna() == False]['Платежей от долга, % или абс.платеж']*df['Месячный доход']
df
```

```python
df.describe()
```

```python
# считаем среднемесячный доход
MeanMonthlyIncome = df['Месячный доход'].mean()

# присываем среднемесячный доход, там где ходод не указан
df['Месячный доход'] = df['Месячный доход'].fillna( MeanMonthlyIncome )

df
```

```python
# вероятность невозврата кредита по полю 'Прогноз просрочки в течении 2 лет'
# от кол-ва иждивенцев и от кол-ва взятых ипотек

# смотрим прогноз по возврату и невозврату для одной и той же группы кредиторов
# например, для группы из 3-х иждевенцев 5 вернули, 10 не вернули, тогда вер-сть невозврата: Р(3) = 10/15 = 0.66666(6)
df[ df['Прогноз просрочки в течении 2 лет'] == 1 ]
```

```python
# прогноз невозварта
df_no_ret_A = df[ df['Прогноз просрочки в течении 2 лет'] == 1 ].groupby('Кол-во иждевенцев в семье').count()['Прогноз просрочки в течении 2 лет']
df_no_ret_A
```

```python
# прогноз возварта
df_ret_A = df[ df['Прогноз просрочки в течении 2 лет'] == 0 ].groupby('Кол-во иждевенцев в семье').count()['Прогноз просрочки в течении 2 лет']
df_ret_A
```

```python
# вер-сть невозврата по кол-ву иждевенцев
# для групп 5,6,8 иждевенцев нет оценки вероятности
df_no_ret_A / (df_ret_A + df_no_ret_A)
```

```python
# аналогично для группы по кол-ву взятых ипотек
df_no_ret_B = df[ df['Прогноз просрочки в течении 2 лет'] == 1 ].groupby('Кол-во ипотек').count()['Прогноз просрочки в течении 2 лет']
df_no_ret_B
```

```python
df_ret_B = df[ df['Прогноз просрочки в течении 2 лет'] == 0 ].groupby('Кол-во ипотек').count()['Прогноз просрочки в течении 2 лет']
df_ret_B
```

```python
# вер-сть невозврата по кол-ву ипотек
# для групп 5,6,8 иждевенцев нет оценки вероятности
df_no_ret_B / (df_ret_B + df_no_ret_B)
```
