Задача 1.
Для выборки N=100, представленной вариационным рядом

$x{_i}=[-1, 0, 1, 2, 3, 4, 5]$\
$n{_i}=[4,  6, 12, 18, 31, 23, 6]$

1. Построить полигон относительных частот и гистограмму накопленных частот.
2. Найти выборочное среднее  и выборочное среднее квадратичное отклонение .
3. Определить доверительный интервал с доверительной вероятностью $\alpha=0.95$  для оценки математического ожидания генеральной совокупности в предположении, что среднее квадратичное уклонение генеральной совокупности равно исправленному выборочному среднему s.
4. Проверить гипотезу о нормальности закона распределения генеральной совокупности, используя критерий Пирсона с уровнем значимости a=0,05.
5. Для вычислений можно использовать язык программирования.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt


# исходные данные
x = pd.Series([-1, 0, 1, 2, 3, 4, 5])
n = pd.Series([4,  6, 12, 18, 31, 23, 6])
```

```python
# собираем данные в один дата фрейм и указываем x - как индекс
df = pd.DataFrame({'x':x, 'n':n}).set_index('x')
df
```

```python
# строим полигон частот
df.plot(grid=True)
```

```python
# т.к. у нас есть распределение частот то строим гистограмму не hist() которая сама считает частоты
# а при помощи bar()
df.plot(y='n', kind='bar', grid=True)
```

```python
print('Выборочное среднее:')
np.sum(df.index * df['n'].values) / df.count().values[0]
```
